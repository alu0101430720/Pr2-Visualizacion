```python
import os
import subprocess
import time
from dotenv import load_dotenv

# 1. Cargar secretos desde el archivo .env
load_dotenv()
os.environ["GITHUB_TOKEN"] = os.getenv("GitHub") 

# 2. AUTODETECCIÓN DE LA CARPETA
if os.path.exists("./structure"):
    # Si ya estás dentro de la carpeta del repositorio
    carpeta_estructura = os.path.abspath("./structure")
elif os.path.exists("./Pr2-Visualizacion/structure"):
    # Si estás una carpeta por encima
    carpeta_estructura = os.path.abspath("./Pr2-Visualizacion/structure")
elif os.path.exists("./Pr2-Visualizacion-Limpia/structure"):
    # Si usaste la carpeta limpia que creamos antes
    carpeta_estructura = os.path.abspath("./Pr2-Visualizacion-Limpia/structure")
else:
    # Si falla, te dirá exactamente dónde estás para que veamos el problema
    raise FileNotFoundError(f"No encuentro la carpeta 'structure'. Tu notebook está ejecutándose en: {os.getcwd()}")

print(f"Usando la carpeta: {carpeta_estructura}")

archivo_log = os.path.abspath("./dagster_logs.txt")
os.environ["PYTHONPATH"] = carpeta_estructura

log_file = open(archivo_log, "w")

# 3. Lanzar Dagster
process = subprocess.Popen(
    ["dagster", "dev", "-f", "definitions.py", "-h", "127.0.0.1", "-p", "3000"],
    cwd=carpeta_estructura,
    stdout=log_file,
    stderr=log_file,
    shell=True 
)

print("Iniciando Dagster...")
time.sleep(8)
log_file.close()

# 4. Leer log
print("\n--- Últimas líneas del log ---")
try:
    with open(archivo_log, "r") as f:
        lineas = f.readlines()
        for linea in lineas[-30:]:
            print(linea.strip())
except FileNotFoundError:
    print("El archivo de log aún no se ha creado.")
```
